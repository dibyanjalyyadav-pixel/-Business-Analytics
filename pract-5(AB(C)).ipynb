```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Set seed for reproducible results
np.random.seed(42)

# Simulate 1000 coin tosses
coin_tosses = np.random.choice(["Heads", "Tails"], size=1000)

print("First 20 Coin Tosses:")
print(coin_tosses[:20])
```

```python
# Calculate frequency
coin_frequency = pd.Series(coin_tosses).value_counts()

# Experimental probability
coin_probability = coin_frequency / 1000

print("Frequency:")
print(coin_frequency)

print("\nExperimental Probability:")
print(coin_probability)
```

```python
theoretical_coin_probability = {
    "Heads": 0.5,
    "Tails": 0.5
}

print("Theoretical Probability:")
print(theoretical_coin_probability)
```

```python
plt.figure(figsize=(6, 5))

plt.bar(
    coin_frequency.index,
    coin_frequency.values
)

plt.title("Frequency of Heads and Tails")
plt.xlabel("Outcome")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    coin_frequency.values,
    labels=coin_frequency.index,
    autopct="%1.1f%%"
)

plt.title("Percentage Distribution of Heads and Tails")

plt.show()
```

```python
import math

n = 10
k = 8
p = 0.5

probability_8_heads = (
    math.comb(n, k)
    * (p ** k)
    * ((1 - p) ** (n - k))
)

print("Probability of getting exactly 8 Heads in 10 tosses:")
print(probability_8_heads)
```

```python
np.random.seed(42)

coin_tosses_10000 = np.random.choice(
    ["Heads", "Tails"],
    size=10000
)

# Frequency
frequency_10000 = pd.Series(
    coin_tosses_10000
).value_counts()

# Experimental probability
probability_10000 = frequency_10000 / 10000

print("Frequency for 10,000 Coin Tosses:")
print(frequency_10000)
print("\nExperimental Probability:")
print(probability_10000)
```

```python
comparison = pd.Series({
    "Head": 0.5,
    "Tail": 0.5
})
comparison.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Comparison of Coin Toss Probabilities")
plt.xlabel("Outcome")
plt.ylabel("Probability")

plt.show()
```
