```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Set seed for reproducible results
np.random.seed(42)

# Simulate 1000 dice rolls
dice_rolls = np.random.randint(1, 7, 1000)

# Display first 20 outcomes
print("First 20 Dice Rolls:")
print(dice_rolls[:20])
```

```python
# Calculate frequency of each outcome
frequency = pd.Series(dice_rolls).value_counts().sort_index()

# Experimental probability
experimental_probability = frequency / 1000

print("Frequency of each outcome:")
print(frequency)

print("\nExperimental Probability:")
print(experimental_probability)
```

```python
theoretical_probability = pd.Series(
    [1/6] * 6,
    index=[1, 2, 3, 4, 5, 6]
)

print("Theoretical Probability:")
print(theoretical_probability)
```

```python
# Probability of each number
print("Probability of getting 1:", experimental_probability[1])
print("Probability of getting 2:", experimental_probability[2])
print("Probability of getting 3:", experimental_probability[3])
print("Probability of getting 4:", experimental_probability[4])
print("Probability of getting 5:", experimental_probability[5])
print("Probability of getting 6:", experimental_probability[6])

# Experimental probabilities of different events
prob_even = np.mean(dice_rolls % 2 == 0)
prob_odd = np.mean(dice_rolls % 2 != 0)
prob_greater_4 = np.mean(dice_rolls > 4)
prob_less_4 = np.mean(dice_rolls < 4)
prob_greater_equal_4 = np.mean(dice_rolls >= 4)

print("\nProbability of Even Number:", prob_even)
print("Probability of Odd Number:", prob_odd)
print("Probability Greater than 4:", prob_greater_4)
print("Probability Less than 4:", prob_less_4)
print("Probability Greater than or Equal to 4:", prob_greater_equal_4)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(frequency.index, frequency.values)

plt.title("Frequency of Dice Outcomes")
plt.xlabel("Dice Outcome")
plt.ylabel("Frequency")
plt.xticks([1, 2, 3, 4, 5, 6])

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    experimental_probability.index,
    experimental_probability.values
)

plt.title("Experimental Probability of Dice Outcomes")
plt.xlabel("Dice Outcome")
plt.ylabel("Probability")
plt.xticks([1, 2, 3, 4, 5, 6])

plt.show()
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    frequency.values,
    labels=frequency.index,
    autopct='%1.1f%%'
)

plt.title("Percentage Distribution of Dice Outcomes")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    experimental_probability.index,
    experimental_probability.values,
    marker='o',
    label='Experimental Probability'
)

plt.plot(
    theoretical_probability.index,
    theoretical_probability.values,
    marker='s',
    label='Theoretical Probability'
)

plt.title("Experimental vs Theoretical Probability")
plt.xlabel("Dice Outcome")
plt.ylabel("Probability")
plt.xticks([1, 2, 3, 4, 5, 6])
plt.legend()

plt.show()
```

```python
even_count = np.sum(dice_rolls % 2 == 0)
odd_count = np.sum(dice_rolls % 2 != 0)

plt.figure(figsize=(6, 5))

plt.bar(
    ["Even", "Odd"],
    [even_count, odd_count]
)

plt.title("Frequency of Even vs Odd Outcomes")
plt.xlabel("Outcome Type")
plt.ylabel("Frequency")

plt.show()
```

```python
event_names = [
    "Even",
    "Odd",
    "Greater than 4",
    "Less than 4"
]

event_probabilities = [
    prob_even,
    prob_odd,
    prob_greater_4,
    prob_less_4
]

plt.figure(figsize=(9, 5))

plt.bar(event_names, event_probabilities)

plt.title("Experimental Probability of Different Events")
plt.xlabel("Events")
plt.ylabel("Experimental Probability")

plt.xticks(rotation=15)

plt.show()
```

```python
# Generate 1000 random numbers from Normal Distribution
normal_data = np.random.normal(loc=50, scale=10, size=1000)

print("First 20 values:")
print(normal_data[:20])
```

```python
# Calculate statistics
mean_value = np.mean(normal_data)
median_value = np.median(normal_data)

# Calculate mode
mode_value = pd.Series(normal_data).mode()

variance_value = np.var(normal_data)
std_value = np.std(normal_data)

print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value.values)
print("Variance:", variance_value)
print("Standard Deviation:", std_value)
```

```python
print("Statistics of Normal Distribution")
print("-------------------------------")

print("Mean:", np.mean(normal_data))
print("Median:", np.median(normal_data))
print("Variance:", np.var(normal_data))
print("Standard Deviation:", np.std(normal_data))
print("Minimum:", np.min(normal_data))
print("Maximum:", np.max(normal_data))
```

```python
plt.figure(figsize=(8, 5))

plt.hist(normal_data, bins=30, edgecolor="black")

plt.title("Histogram of Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

plt.boxplot(normal_data)

plt.title("Box Plot of Normal Distribution")
plt.ylabel("Values")

plt.show()
```

```python
percentage_40_60 = np.mean(
    (normal_data >= 40) & (normal_data <= 60)
) * 100

print("Percentage of values between 40 and 60:",
      percentage_40_60, "%")
```

```python
percentage_30_70 = np.mean(
    (normal_data >= 30) & (normal_data <= 70)
) * 100

print("Percentage of values between 30 and 70:",
      percentage_30_70, "%")
```

```python
exponential_data = np.random.exponential(
    scale=2,
    size=1000
)

print("First 20 values:")
print(exponential_data[:20])
```

```python
plt.figure(figsize=(8, 5))

plt.hist(exponential_data, bins=30, edgecolor="black")

plt.title("Histogram of Exponential Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

plt.boxplot(exponential_data)

plt.title("Box Plot of Exponential Distribution")
plt.ylabel("Values")

plt.show()
```

```python
Q1 = np.percentile(exponential_data, 25)
Q3 = np.percentile(exponential_data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = exponential_data[
    (exponential_data < lower_bound) |
    (exponential_data > upper_bound)
]

print("Possible Extreme Values:")
print(outliers)
```

```python
percentage_greater_2 = np.mean(
    exponential_data > 2
) * 100

print("Percentage of values greater than 2:",
      percentage_greater_2, "%")
```

```python
percentage_greater_5 = np.mean(
    exponential_data > 5
) * 100

print("Percentage of values greater than 5:",
      percentage_greater_5, "%")
```
