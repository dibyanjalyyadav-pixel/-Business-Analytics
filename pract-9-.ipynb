```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.stats.proportion import proportions_ztest
```

```python
#1. Load the A/B testing dataset
df = pd.read_csv("ab_testing.csv")
```

```python
# 2. Display the data
print("A/B Testing Data:")
print(df)
```

```python
# 3. Calculate conversion rate for Group A
group_a = df[df["Group"] == "A"].iloc[0]
```

```python
# 3. Calculate conversion rate for Group A
group_a = df[df["Group"] == "A"].iloc[0]

conversion_rate_a = (
    group_a["Conversions"] / group_a["Users"]
)

print("\nConversion Rate of Group A:")
print(conversion_rate_a)
```

```python
# 4. Calculate conversion rate for Group B
group_b = df[df["Group"] == "B"].iloc[0]

conversion_rate_b = (
    group_b["Conversions"] / group_b["Users"]
)

print("\nConversion Rate of Group B:")
print(conversion_rate_b)
```

```python
# Convert into percentage
print("\nGroup A Conversion Rate:")
print(conversion_rate_a * 100, "%")

print("\nGroup B Conversion Rate:")
print(conversion_rate_b * 100, "%")
```

```python
# 5. Compare conversion rates
print("\nComparison:")
if conversion_rate_a > conversion_rate_b:
    print("Group A has a higher conversion rate.")
elif conversion_rate_b > conversion_rate_a:
    print("Group B has a higher conversion rate.")
else:
    print("Both groups have the same conversion rate.")
```

```python
# 6. Null Hypothesis
print("\nNull Hypothesis:")
print("H0: pA = pB")
```

```python
# 7. Alternative Hypothesis
print("\nAlternative Hypothesis:")
print("H1: pA ≠ pB")
```

```python
# 8. Significance level
alpha = 0.05

print("\nSignificance Level:")
print(alpha)
```

```python
# 9. Perform two-proportion test
conversions = np.array([
    group_a["Conversions"],
    group_b["Conversions"]
])

users = np.array([
    group_a["Users"],
    group_b["Users"]
])

z_statistic, p_value = proportions_ztest(
    conversions,
    users
)
```

```python
# 10. Display p-value
print("\nZ-Statistic:")
print(z_statistic)

print("\nP-Value:")
print(p_value)
```

```python
# 11. Compare p-value with 0.05
print("\nComparison:")

if p_value < alpha:
    print("p-value < 0.05")
else:
    print("p-value >= 0.05")
```

```python
# 12. Decision
print("\nDecision:")

if p_value < alpha:
    print("Reject the Null Hypothesis (H0)")
else:
    print("Do Not Reject the Null Hypothesis (H0)")
```

```python
# 13. Create bar chart
groups = ["A", "B"]
conversion_rates = [
    conversion_rate_a * 100,
    conversion_rate_b * 100
]

plt.bar(groups, conversion_rates)

plt.xlabel("Group")
plt.ylabel("Conversion Rate (%)")
plt.title("A/B Testing - Conversion Rate Comparison")

plt.ylim(0, max(conversion_rates) + 5)

# Display values on bars
for i, value in enumerate(conversion_rates):
    plt.text(i, value + 0.5, f"{value:.1f}%",
             ha="center")

plt.show()
```

```python
# 14. Conclusion
print("\nConclusion:")

if p_value < alpha:
    print("The p-value is less than 0.05.")
    print("Therefore, we reject the null hypothesis.")
    print("There is a statistically significant difference")
    print("between the conversion rates of Group A and Group B.")
else:
    print("The p-value is greater than or equal to 0.05.")
    print("Therefore, we do not reject the null hypothesis.")
    print("There is no statistically significant difference")
    print("between the conversion rates of Group A and Group B.")
```
